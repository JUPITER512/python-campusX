```python
# dictionary comprehension
d3={x:x**2 for x in range(1,11)}
d3
```

```python
distances_km={"lahore":1000,"karachi":2000}
distance_miles={key:value*1.69 for key,value in distances_km.items()}
print(distance_miles)
```
