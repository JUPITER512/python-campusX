```python
# positional arguments
# Matched arguments to parameters by order
print(power(2,3))
print(power(3,2))
```

```python
# keyword argument
# Matched by name, so order doesn't matter.

# match the arguments by name the of parameters
power(b=2,a=4)
```

```python
# *args
# allows us to pass a variable number of non-keyword arguments to a function.
# Python packs the extra values into a tuple.
# *args=variable arguments
# (*)variable_name means astrike zaruri ha 
def multiply(*args):
    print(args)
    product=1
    for i in args:
        product*=i
        print(product)
    return product

multiply(1,2,3,4,5,6)
```

```python
# (**kwargs) keyword arguments
# **kwargs allows us to pass any number of keyword arguments
# keyword arguments mean that they contain a key value pair like a python dictionary
# only ("**") important
def display(**kwargs):
    for key,value in kwargs.items():
        print(f"{key} --> {value}")

display(pakistan="lahore",india="bombay",nepal="kathmandu")
```

```python
# higher order function
# function return a function 
# function receive a function as an argument
def square(x):
    return x**2
def transform (func,l):
    output=[]
    for i in l:
        output.append(func(i))
    print(output)
L=[1,2,3,4,5]

transform(lambda x:x**2,L)
transform(lambda x:x**3,L)
```

```python
# map
users = [
    {'name': 'Rahul', 'age': 45, 'gender': 'male'},
    {'name': 'Nitish', 'age': 33, 'gender': 'male'},
    {'name': 'Ankita', 'age': 50, 'gender': 'female'},
]

list(map(lambda user : user['name'],users))
list(map(lambda user : user['age'],users))
list(map(lambda user : user['gender'],users))
```
