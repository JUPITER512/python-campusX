```python
s1={1,2,3,4,5}
s2={4,5,6,7,8}
# set operations
# union
s1 | s2
# intersection
s1 & s2
# difference
s1 - s2
# symmetric difference(^)
s1^s2
# membership test
2 in s1
1 not in s2
# iteration
for i in s2:
    print(i)
```

```python
s={1,3,2,1,4,10,20,1,4}
len(s)
sum(s)
min(s)
max(s)
sorted(s,reverse=True)
```

```python
s1={1,2,3,4,5}
s2={6,7,8,9,10}
s1.union(s2)

# union update
s1.update(s2)#permanently update the s1 set
print(s1)
print(s2)
```

```python
# difference and difference update
s1.difference(s2)
s1.difference_update(s2)
print(s1)
print(s2)
```
