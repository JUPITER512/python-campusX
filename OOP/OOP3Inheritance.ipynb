```python
class Phone:
    def __init__(self,price,brand,camera):
        print("Inside the phone constructor")
        self.__price=price
        self.brand=brand
        self.camera=camera

class Smartphone(Phone):
    def __init__(self, price, brand, camera,os, ram):
        print("Inside the smartphone constructor")
        super().__init__(price,camera,brand)
        self.os=os
        self.ram=ram
        print("Inside the smartphone constructor")

s=Smartphone(2000,"Samsung",12,"Andriod",2)
print(s.os)
print(s.brand)
```
