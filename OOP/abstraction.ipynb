```python
# from abc import ABC,abstract methods
from abc import ABC,abstractmethod
class BankAPP(ABC):
    def database(self):
        print("Connected to database")
    @abstractmethod
    def security(self):
        pass
```

```python
class MobileApp(BankAPP):
    def mobile_login(self):
        print("Login")
    def security(self):
        print("Mobile Security")
```

```python
mobile=MobileApp()
```

```python
mobile.security()
```
